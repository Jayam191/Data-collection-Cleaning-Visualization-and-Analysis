```python
#task 4 

import numpy as np

ratings = np.array([
    [4, 1, 0],
    [1, 3, 1],
    [0, 1, 2]
])

inverse = np.linalg.inv(ratings)
determinant = np.linalg.det(ratings)
eigenvalues, eigenvectors = np.linalg.eig(ratings)

print("Inverse:")
print(inverse)

print("\nDeterminant:")
print(determinant)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
#task 5 

import numpy as np

orders = np.array([
    [10, 20, 30, 40, 50, 60],
    [15, 25, 35, 45, 55, 65]
])

print("Original shape:", orders.shape)

# Reshape to 3 × 4
reshaped = orders.reshape(3, 4)
print("\nReshaped:")
print(reshaped)

# Flatten
flattened = reshaped.flatten()
print("\nFlattened:")
print(flattened)

# Split into two equal parts
parts = np.split(flattened, 2)

print("\nPart 1:")
print(parts[0])

print("Part 2:")
print(parts[1])

# Stack vertically
stacked = np.vstack(parts)

print("\nStacked:")
print(stacked)
```
