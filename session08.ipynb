```python
#task 4 

import pandas as pd

df = pd.DataFrame({
    "username": [
        "insta_queen",
        "insta-queen",
        "instaqueen",
        "john123",
        "insta_queen",
        "insta-queen"
    ]
})

print("Before:")
print(df)


df["username"] = df["username"].replace({
    "insta_queen": "instaqueen",
    "insta-queen": "instaqueen"
})

print("\nAfter:")
print(df)


```

```python
#task 5 

import pandas as pd

df = pd.DataFrame({
    "payment_status": [
        "Yes",
        " yes ",
        "Y",
        "No",
        " no ",
        "N",
        " YES ",
        " n "
    ]
})

print("Before:")
print(df)

df["payment_status"] = df["payment_status"].str.strip().str.lower()

df["payment_status"] = df["payment_status"].map({
    "yes": 1,
    "y": 1,
    "no": 0,
    "n": 0
})

print("\nAfter:")
print(df)
```
