```python
#task 1 

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

delivery_times = np.random.randint(15, 61, 50)

sns.histplot(
    delivery_times,
    bins=8,
    kde=True
)

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Zomato Delivery Times")

plt.show()
```

```python
#task 2 

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.boxplot(
    x="day",
    y="total_bill",
    data=df
)

plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")
plt.title("Distribution of Total Bills by Day")

plt.show()
```

```python
#task 3 


import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)

teams = ["CSK", "MI", "RCB", "KKR", "SRH", "DC", "PBKS", "RR"]

data = []

for team in teams:
    scores = np.random.normal(170, 30, 20)
    scores = np.clip(scores, 80, 250)

    for score in scores:
        data.append([team, score])

df = pd.DataFrame(data, columns=["Team", "Runs"])

sns.set_theme(style="darkgrid")

sns.violinplot(
    x="Team",
    y="Runs",
    data=df
)

plt.xlabel("IPL Team")
plt.ylabel("Runs Scored")
plt.title("IPL Run Distribution by Team")

plt.show()
```

```python
#task  4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

genres = [
    "Pop", "Rock", "Hip-Hop", "Pop", "EDM",
    "Rock", "Pop", "Hip-Hop", "EDM", "Pop",
    "Rock", "Hip-Hop", "Pop", "EDM", "Rock",
    "Pop", "Hip-Hop", "EDM", "Rock", "Pop",
    "Pop", "Rock", "Hip-Hop", "EDM", "Pop",
    "Rock", "Hip-Hop", "EDM", "Pop", "Rock",
    "Hip-Hop", "Pop", "EDM", "Rock", "Pop",
    "Hip-Hop", "EDM", "Rock", "Pop", "Hip-Hop"
]

df = pd.DataFrame({"Genre": genres})

sns.countplot(x="Genre", data=df)

plt.xlabel("Music Genre")
plt.ylabel("Number of Songs")
plt.title("Number of Spotify Songs per Genre")

plt.show()
```
