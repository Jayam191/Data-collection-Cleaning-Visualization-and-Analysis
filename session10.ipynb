```python
#task 1

import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

steps = [5000, 6500, 7200, 6000, 8000, 9500, 7000]

plt.plot(days, steps, marker="o")

plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.title("Daily Steps - Last 7 Days")

plt.xticks(rotation=45)

plt.tight_layout()


plt.show()
```

```python
#task 3 

import matplotlib.pyplot as plt

apps = ["Swiggy", "Zomato", "Domino's"]
orders = [12, 8, 5]

colors = ["orange", "red", "blue"]

plt.bar(apps, orders, color=colors, label="Orders")

plt.xlabel("Food Delivery App")
plt.ylabel("Number of Orders")
plt.title("Food Delivery Orders - Last Month")

plt.legend()

plt.show()
```

```python
#task 4 

import matplotlib.pyplot as plt

durations = [
    15, 20, 25, 30, 35,
    18, 22, 27, 32, 40,
    45, 50, 28, 35, 42,
    55, 60, 38, 25, 30
]

plt.hist(durations, bins=5, color="purple")

plt.xlabel("Listening Duration (minutes)")
plt.ylabel("Number of Sessions")
plt.title("Spotify Listening Session Durations")

plt.show()
```

```python
#task 5 



import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

screen_time = [2, 3, 2.5, 4, 3.5, 5, 4.5]

liked_posts = [10, 15, 12, 20, 18, 25, 22]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(days, screen_time, marker="o")

axes[0].set_title("Daily Instagram Screen Time")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Screen Time (hours)")

axes[1].bar(days, liked_posts, color="orange")

axes[1].set_title("Instagram Posts Liked")
axes[1].set_xlabel("Day")
axes[1].set_ylabel("Number of Posts Liked")

plt.tight_layout()

plt.savefig("social_media_usage.png")

# Display the figure
plt.show()
```
