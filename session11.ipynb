```python
#task 1 

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [20, 35, 30, 45, 40]

products = ["Pizza", "Burger", "Pasta", "Biryani"]
orders = [30, 25, 20, 35]

age = [18, 20, 22, 25, 28, 30, 32, 35]
salary = [20000, 25000, 28000, 35000, 40000, 45000, 50000, 60000]

apps = ["Swiggy", "Zomato", "Dominos"]
usage = [40, 35, 25]


fig, axes = plt.subplots(2, 2, figsize=(12, 8))



axes[0, 0].plot(days, sales, marker="o")
axes[0, 0].set_title("Daily Sales")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Sales")



axes[0, 1].bar(products, orders)
axes[0, 1].set_title("Orders by Product")
axes[0, 1].set_xlabel("Product")
axes[0, 1].set_ylabel("Orders")


axes[1, 0].scatter(age, salary)
axes[1, 0].set_title("Age vs Salary")
axes[1, 0].set_xlabel("Age")
axes[1, 0].set_ylabel("Salary")


axes[1, 1].pie(usage, labels=apps, autopct="%1.1f%%")
axes[1, 1].set_title("Food Delivery App Usage")


plt.tight_layout()


# plt.savefig("four_charts.png")

plt.show()
```

```python
#task 2 

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Domino's"]
delivery_times = [32, 28, 35]


colors = ["red", "orange", "blue"]


bars = plt.bar(
    apps,
    delivery_times,
    color=colors,
    edgecolor="black",
    linewidth=2
)

bars[1].set_linestyle("--")   
bars[2].set_linestyle(":")    

plt.xlabel("Food Delivery App")
plt.ylabel("Average Delivery Time (minutes)")
plt.title("Average Delivery Time Comparison")

plt.show()
```

```python
#task 3 

import matplotlib.pyplot as plt

influencers = ["A", "B", "C", "D", "E"]

followers = [120000, 250000, 180000, 350000, 200000]
daily_posts = [2, 4, 3, 5, 2.5]

fig, ax1 = plt.subplots()

ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")


ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")


plt.title("Instagram Followers vs Average Daily Posts")

plt.show()
```

```python
#task 4 

import matplotlib.pyplot as plt

movies = [
    "Dhurandhar: The Revenge",
    "Border 2",
    "Cocktail 2",
    "Ikkis",
    "Mardaani 3"
]

tickets_sold = [18, 9, 7, 5, 4]

ratings = [8.2, 5.8, 6.5, 6.1, 6.0]


plt.scatter(
    tickets_sold,
    ratings,
    color="red",
    marker="o",
    s=100
)


for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], ratings[i]),
        xytext=(5, 8),
        textcoords="offset points"
    )


plt.xlabel("Tickets Sold (millions)")
plt.ylabel("IMDb Rating")
plt.title("Tickets Sold vs IMDb Ratings")


plt.grid(True, alpha=0.3)

plt.show()
```

```python
#task 5 

import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

sales = [
    120, 135, 128, 145, 150, 160,
    155, 170, 180, 450, 200, 190
]

fig, ax = plt.subplots()

ax.plot(months, sales, marker="o")

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    "Big Billion Days",
    xy=(months[max_index], max_sales),
    xytext=(0, 40),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Crore)")
ax.set_title("Flipkart Monthly Sales")

plt.tight_layout()
plt.show()
```
