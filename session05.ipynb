```python
#task 1 

import pandas as pd
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:1234@localhost:3306/zomato"
)

with engine.connect() as connection:
    print("MySQL connection successful!")

    df = pd.read_sql("SELECT * FROM restaurants", connection)

print(df.head())    
```

```python
#task 2 

with engine.connect() as connection:
    df = pd.read_sql_query(
        "SELECT name, rating FROM movies WHERE rating > 8",
        connection
    )

print(df)
```

```python
#task 4 

import pandas as pd
from pathlib import Path

orders_path = Path("orders05.csv")
users_path = Path("users05.csv")


orders_df = pd.read_csv(orders_path)
users_df = pd.read_csv(users_path)

df = pd.merge(orders_df, users_df, on="user_id")


print(df[["username", "amount"]])
```

```python
# task 5

import pandas as pd

today_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "item": ["Pizza", "Burger", "Pasta"],
    "price": [250, 180, 220]
})

yesterday_orders = pd.DataFrame({
    "order_id": [98, 99, 100],
    "item": ["Sandwich", "Pizza", "Biryani"],
    "price": [150, 300, 280]
})

combined_orders = pd.concat(
    [today_orders, yesterday_orders],
    ignore_index=True
)

print(combined_orders)
```
