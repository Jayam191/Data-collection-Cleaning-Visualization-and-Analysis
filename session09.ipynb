```python
#task 2 

import pandas as pd

df = pd.read_csv("flipkart_orders09.csv")

df["order_date"] = pd.to_datetime(df["order_date"])

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["weekday"] = df["order_date"].dt.day_name()

print(df)
```

```python
#task 3 

import pandas as pd

df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "order_date": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2025-01-15 18:45:00",
        "2025-03-20 11:20:00",
        "2025-06-10 16:30:00"
    ]
})
df["order_date"] = pd.to_datetime(df["order_date"])

df = df.set_index("order_date")

weekly_orders = df.resample("W").size()

print(weekly_orders)
```

```python
#task 4 

import pandas as pd
df = pd.DataFrame({
    "post_id": [101, 102, 103, 104, 105],
    "posted_at": [
        "2024-06-01 10:00:00",
        "2024-06-02 14:30:00",
        "2024-06-03 18:45:00",
        "2024-06-04 09:15:00",
        "2024-06-05 20:00:00"
    ]
})
df["posted_at"] = pd.to_datetime(df["posted_at"], utc=True)
df["posted_at"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")
print(df["posted_at"].head())
```

```python
#task 5 

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "order_date": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2025-01-15 18:45:00",
        "2025-03-20 11:20:00",
        "2025-06-10 16:30:00"
    ]
})

df["order_date"] = pd.to_datetime(df["order_date"])

df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

print(df)
```
