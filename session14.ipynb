```python
#task 2

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "rating": [5, 4, 5, 3, 2, 5, 4, 1, 3, 5,
               4, 5, 2, 3, 4, 5, 1, 4, 3, 5],
    "category": ["Mobile", "Laptop", "Mobile", "Headphones", "Shoes",
                 "Laptop", "Mobile", "Shoes", "Headphones", "Mobile",
                 "Laptop", "Shoes", "Mobile", "Laptop", "Headphones",
                 "Mobile", "Shoes", "Laptop", "Mobile", "Headphones"]
}

df = pd.DataFrame(data)

sns.countplot(
    data=df,
    x="rating"
)

plt.xlabel("Rating (Stars)")
plt.ylabel("Number of Reviews")
plt.title("Number of Flipkart Reviews by Rating")

plt.show()
```

```python
#task 3 

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "average_cost_for_two": [200, 500, 800, 1200, 350, 600, 1500, 250, 700, 1000],
    "user_rating": [3.5, 4.0, 4.2, 4.5, 3.2, 3.8, 4.7, 3.0, 4.1, 4.4]
}

df = pd.DataFrame(data)

sns.scatterplot(
    data=df,
    x="average_cost_for_two",
    y="user_rating"
)

plt.xlabel("Average Cost for Two (₹)")
plt.ylabel("User Rating")
plt.title("Zomato Restaurant Cost vs User Rating")

plt.show()
```

```python
#task 4 

import pandas as pd

data = {
    "genre": [
        "Action", "Comedy", "Drama", "Action", "Comedy",
        "Drama", "Horror", "Action", "Horror", "Comedy"
    ],
    "box_office_collection": [
        250, 180, 150, 300, 200,
        170, 120, 350, 140, 220
    ]
}

df = pd.DataFrame(data)

result = (
    df.groupby("genre")["box_office_collection"]
      .mean()
      .sort_values(ascending=False)
)

print(result)
```
