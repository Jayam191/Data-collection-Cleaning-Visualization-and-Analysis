```python
import pandas as pd

df = pd.DataFrame({
    "restaurant": [
        "Spice Garden",
        "Urban Tadka",
        "Food Junction",
        "Royal Treat",
        "Cafe Aroma",
        "Green Leaf"
    ],
    "rating": [None, 4.5, None, 4.3, None, 3.6]
})

print("Before:")
print(df)

df["rating"] = df["rating"].ffill()
df["rating"] = df["rating"].bfill()

print("\nAfter:")
print(df)
```
